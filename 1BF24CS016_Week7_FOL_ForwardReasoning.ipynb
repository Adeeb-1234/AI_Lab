```python
KB = {
    ("American", "Robert"),
    ("Missile", "T1"),
    ("Owns", "A", "T1"),
    ("Enemy", "A", "America")
}


rules = [
    ([("Missile", "x")],
     ("Weapon", "x")),

    ([("Enemy", "x", "America")],
     ("Hostile", "x")),

    ([("Missile", "x"), ("Owns", "A", "x")],
     ("Sells", "Robert", "x", "A")),

    ([("American", "p"), ("Weapon", "q"),
      ("Sells", "p", "q", "r"), ("Hostile", "r")],
     ("Criminal", "p"))
]


def match(pattern, fact, substitution):
    substitution = substitution.copy()

    if pattern[0] != fact[0] or len(pattern) != len(fact):
        return None

    for p, f in zip(pattern[1:], fact[1:]):
        if p.islower():  # Variable
            if p in substitution and substitution[p] != f:
                return None
            substitution[p] = f
        elif p != f:
            return None

    return substitution



def find_matches(conditions, KB):
    substitutions = [{}]

    for condition in conditions:
        new_substitutions = []

        for sub in substitutions:
            for fact in KB:
                result = match(condition, fact, sub)

                if result is not None:
                    new_substitutions.append(result)

        substitutions = new_substitutions

    return substitutions



def substitute(expression, sub):
    return tuple(sub.get(x, x) for x in expression)



def forward_chaining(KB, rules, query):

    while True:
        new_facts = set()

        for conditions, conclusion in rules:

            substitutions = find_matches(conditions, KB)

            for sub in substitutions:
                fact = substitute(conclusion, sub)

                if fact not in KB:
                    new_facts.add(fact)

                if fact == query:
                    return True

        if not new_facts:
            return False

        KB.update(new_facts)

        print("\nNew facts generated:")
        for fact in new_facts:
            print(fact)



query = ("Criminal", "Robert")

if forward_chaining(KB, rules, query):
    print("\nQuery Proved:", query)
else:
    print("\nQuery Not Proved")

```
