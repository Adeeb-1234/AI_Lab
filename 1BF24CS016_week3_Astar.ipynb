```python
import heapq

# Initial and goal states
START = "283164705"
GOAL = "123804765"


# -------------------------------------------------
# Generate all possible next states
# -------------------------------------------------
def get_neighbors(state):
    neighbors = []

    zero_pos = state.index("0")
    row = zero_pos // 3
    col = zero_pos % 3

    moves = [
        (-1, 0),  # Up
        (1, 0),   # Down
        (0, -1),  # Left
        (0, 1)    # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_pos = new_row * 3 + new_col

            state_list = list(state)

            # Swap blank with the neighboring tile
            state_list[zero_pos], state_list[new_pos] = \
                state_list[new_pos], state_list[zero_pos]

            new_state = ''.join(state_list)

            neighbors.append(new_state)

    return neighbors


# -------------------------------------------------
# Heuristic 1: Number of misplaced tiles
# -------------------------------------------------
def misplaced_tiles(state, goal):

    count = 0

    for i in range(9):

        # Don't count the blank
        if state[i] != '0' and state[i] != goal[i]:
            count += 1

    return count


# -------------------------------------------------
# Heuristic 2: Manhattan Distance
# -------------------------------------------------
def manhattan_distance(state, goal):

    distance = 0

    for tile in "12345678":

        current_pos = state.index(tile)
        goal_pos = goal.index(tile)

        current_row = current_pos // 3
        current_col = current_pos % 3

        goal_row = goal_pos // 3
        goal_col = goal_pos % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance


# -------------------------------------------------
# Print a puzzle state
# -------------------------------------------------
def print_state(state):

    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])


# -------------------------------------------------
# A* Search
# -------------------------------------------------
def a_star(start, goal, heuristic):

    # Priority queue:
    # (f, g, state, path)
    pq = []

    g_start = 0
    h_start = heuristic(start, goal)
    f_start = g_start + h_start

    heapq.heappush(
        pq,
        (f_start, g_start, start, [start])
    )

    visited = set()

    while pq:

        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        h = f - g

        print("\nState:")
        print_state(state)

        print("g =", g)
        print("h =", h)
        print("f =", f)

        # Goal reached
        if state == goal:
            print("\nGoal reached!")
            print("Number of moves =", g)
            return path

        # Generate children
        for child in get_neighbors(state):

            if child not in visited:

                new_g = g + 1
                new_h = heuristic(child, goal)
                new_f = new_g + new_h

                heapq.heappush(
                    pq,
                    (new_f, new_g, child, path + [child])
                )

    return None


# =================================================
# CASE 1: MISPLACED TILES
# =================================================

print("=" * 50)
print("A* USING MISPLACED TILES")
print("=" * 50)

path1 = a_star(
    START,
    GOAL,
    misplaced_tiles
)


# =================================================
# CASE 2: MANHATTAN DISTANCE
# =================================================

print("\n\n")
print("=" * 50)
print("A* USING MANHATTAN DISTANCE")
print("=" * 50)

path2 = a_star(
    START,
    GOAL,
    manhattan_distance
)
```
