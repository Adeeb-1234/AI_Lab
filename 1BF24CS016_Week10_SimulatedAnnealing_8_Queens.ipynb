```python
import random
import math

# Calculate attacking pairs
def cost(board):
    attacks = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if abs(board[i] - board[j]) == abs(i - j):
                attacks += 1
    return attacks

# Simulated Annealing
def solve():
    board = [random.randint(0, 7) for _ in range(8)]

    T = 100
    alpha = 0.99

    while T > 0.1:
        old_cost = cost(board)

        if old_cost == 0:
            break

        new_board = board.copy()

        col = random.randint(0, 7)
        new_board[col] = random.randint(0, 7)

        new_cost = cost(new_board)

        if new_cost < old_cost:
            board = new_board

        else:
            p = math.exp((old_cost - new_cost) / T)

            if random.random() < p:
                board = new_board

        T = T * alpha

    if cost(board) == 0:
        print("Solution found:")
        print(board)
    else:
        print("No solution found")

solve()
```
